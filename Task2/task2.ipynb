## Importing Modules

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, KFold
```

## Reading the file

```python
df = pd.read_csv('../chicago_crime_dataset.csv')
```

## Preprocessing and filtering the data

```python
df = df.dropna()
df['Date'] = pd.to_datetime(df['Date'])
df['timestamp'] = df['Date'].dt.strftime('%H:%M')
df = df[df['timestamp'] != '00:00']
```

## Splitting Features into Given and Target Features

```python
final_features = [ 'Primary Type', 'Location Description', 'Domestic',  'District',  'Year','timestamp','Arrest']

df[final_features].head(10)
df[final_features].info()
```

```python
X = df[final_features].drop('Arrest', axis=1)
y = df['Arrest']
```

## Preprocessing and encoding

```python
preprocessor = ColumnTransformer([
    ('onehot', OneHotEncoder(handle_unknown='ignore'), ['Primary Type', 'Location Description', 'Domestic', 'timestamp'])
], remainder='passthrough')
```

## Model Initialisation

```python
model1 = RandomForestClassifier()
model2 = DecisionTreeClassifier()
```

## Pipeline

```python
pipeline1 = Pipeline([('preprocessor', preprocessor), ('classifier', model1)])
pipeline2 = Pipeline([('preprocessor', preprocessor), ('classifier', model2)])
```

## K Fold Cross Validation

```python
scores_model1 = cross_val_score(pipeline1, X, y, cv=5)
scores_model2 = cross_val_score(pipeline2, X, y, cv=5)
```

## Printing and comparing scores



```python
print("RandomForestClassifier : ", scores_model1.mean())
print("DecisionTreeClassifier : ", scores_model2.mean())
```

## Final View

DecisionTreeClassifer and RandomForestClassifer models have been used to compare on the given dataset using K-Flip Cross Validation.

As per the observation on the mean scores of both the models, RandomForestClassifer has been more accurate than DecisionTreeCLassifier, with their accuracies being 79.868% and 77.118% respectively
